***Key Technical Guidlines***

1. Bambu PA6-CF Thermal Limits & Moisture: 

While Bambu PA6-CF lists a high Heat Deflection Temperature (186°C), HDT only reflects short-term mechanical load in dry conditions. Nylon 6 naturally absorbs atmospheric humidity. When heated in an oven during a prepreg cure (e.g., 90–120°C for our XPREG® XC110), trapped water molecules vaporize, causing severe outgassing. This creates surface voids in the carbon fiber and delamination within the 3D print itself.

2. 2-Piece Matched Mold vs. Gloss Finish:

Prepreg carbon fiber cannot be properly consolidated using standard mechanical clamping in a rigid 2-piece mold. Unlike liquid resin, prepreg needs uniform atmospheric compaction (via a vacuum bag) to collapse plies and bleed out air. Clamping rigid mold halves together leads to resin bridging, dry spots, and massive voiding.
Furthermore, 3D print layer lines will "telegraph" directly onto your carbon fiber part. To achieve a high-gloss finish, the tool surface must be skim-coated with an epoxy coating resin (like our XCR Coating Resin), flatted up to 1200+ grit, polished, and treated with CR1 Easy-Lease Chemical Release Agent.

3. Recommended Production Workflow:

A single-sided mold with a vacuum bag is the standard, reliable method for complex geometries like welding hoods. The most cost-effective and durable approach is a two-stage process:
- Print a Pattern (Plug): 3D print your square hood design using PLA, PETG, or PA6-CF.
- Surface Prep: Sand the 3D print, coat it with XCR Epoxy Coating Resin, and polish it to a mirror finish.
- Make a Composite Mold: Cast a dedicated female mold off your printed pattern using our EG160 High-Temp Epoxy Tooling Gelcoat and EMP160 Tooling Paste (or XPREG® XT135 Tooling Prepreg).
- Cure Prepreg Parts: Lay up your prepreg into this composite mold, vacuum bag it, and cure it in an oven at 120°C.

## Short Comments

Based on the above, we will need to pull the carbon fiber prepregs in a vacuum while baking. This is whether we use a 2 part sandwich mold, one part mold, or another method. We are kinda SOL on that part, but there are still some more cost effective methods of getting the first prototype out. Below are three alternate molding methods: 

1. High-Temp Silicone Rubber Mold: 
- Process: 3D print a master plug of the hood in cheap PLA or PETG. Sand and paint it smooth, then pour high-temperature silicone (such as Smooth-On Mold Max 60, rated to 290°C) over the plug to cast a flexible female mold.
- Benefits: Silicone is 100% airtight, requires no mold release agent, and its flexibility makes popping out deep, complex welding hood shapes vastly easier than rigid tooling.

2. 3D-Printed Master Plug + Fiberglass/Epoxy Tooling (Best for Multiple Hoods): 
- Process: 3D print the hood pattern in PLA/PETG, seal layer lines with coating epoxy resin, sand to a mirror finish, and apply chemical release agent. Lay up high-temp epoxy tooling gelcoat backed with budget fiberglass cloth (instead of expensive carbon fiber) over the pattern.
- Benefits: Fiberglass provides the exact same structural and thermal stability as carbon fiber tooling for prepreg, but at a fraction of the raw material cost.

3. Direct High-Temperature Polycarbonate (PC) Tooling:
- Process: Print the mold directly using Polycarbonate (PC) or PC-CF (which has a true glass transition temperature of 140°C–150°C). Coat the mold in high-temp surface primer, flat-sand to 1200 grit, seal, and bake.
- Benefits: Higher true thermal resistance than Nylon or PET, preventing vacuum bag collapse during a 120°C prepreg cure.

# ***Making a Composite Mold from a 3D Print***
Disclaimer: all instructions and processes described in this markdown cell belong to Easy Composites US. The link to this description is attached: [Making a Composite Mold from a 3D Print](https://www.easycomposites.us/learning/making-a-composites-mold-from-a-3d-print)


### ***Introduction:*** 

There are many ways that patterns can be produced ranging from simple hand shaping by eye in materials such as wood and foam through to 5 axis CNC machining epoxy or polyurethane modelboard, in this tutorial we will be focusing on the use of FDM printers to create the initial pattern form. 3D printing technology makes for a cost-effective pattern making solution that is both fast and clean and is particularly well suited to development of small composite parts.

We will be taking an FDM print with the barriers already modelled and coating this with an epoxy coating that will both seal and smoothen the surface. After flatting and polishing this pattern can then be used to create a mold using any conventional room temperature curing mold making method. In this project we will use the [high-temperature capable EG160 high temperature epoxy gelcoat](https://www.easycomposites.us/eg160-high-temperature-epoxy-tooling-gel-coat) and [EMP160 high temperature epoxy laminating paste](https://www.easycomposites.us/emp160-high-temperature-epoxy-molding-paste).

Once the molds have been produced we will go on to manufacture the component using the [XPREG XC110 out of autoclave carbon fiber prepreg](https://www.easycomposites.us/xc110-210g-22-twill-3k-prepreg-carbon-fiber).

### ***Material Compatability:***

There are a wide range of print materials that can be used for pattern making however we strongly recommend using PLA or ABS, not only are these to most readily available print filaments but the resulting print will also offer good adhesion with the [XCR Epoxy Coating Resin](https://www.easycomposites.us/xcr-epoxy-coating-resin) that is used to seal and smoothen the print. Other print materials my offer a reduced bond so a small test for compatibility should be conducted if you plan to use a different type.

The pattern will then be coated with XCR coating resin, this is a specifically developed coating resin and offers both excellent substrate adhesion and surface levelling and gloss properties. Once cured the XCR coating resin can be used to produce molds from all conventional mold making materials including epoxy, polyester and vinylester.

The next compatibility question is that with the chosen mold making system; this must be compatible with the final production technique, in this project we have used the high temperature epoxy EG160/EMP160 combination as we needed the mold to be able to withstand elevated temperature processing for prepreg, however for ambient temperature use molds for wet-lay-up, vacuum-bagging or resin infusion, systems such as Uni-Mold or EG60/EMP60 could be used. The main type of tooling that is not recommended to be taken from 3d printed patterns is prepreg tooling such as XT135 as the elevated temperature and vacuum is liable to significantly distort or collapse the 3d print.

### ***Materials & Equipment Needed:***

- Printing Equipment and Materials

You will need a properly configured and setup FDM printer, in this project we have used the excellent Ultimaker S5 which with its pre-sets and intuitive user interface makes 3D printing easy, however any printer that is properly configured can be used. If you are looking for advice on 3d printing we would be happy to recommend our suppliers Dynamism.

We recommend using PLA filament, not only it this one of the most readily available and reliably printed materials but it also offer good adhesion to the coating resin.

- Pattern Coating and Finishing Materials

XCR Epoxy Coating Resin is specifically designed for high quality coating applications and has been proven to offer unbeatable bond to the print, and surface finish when coating FDM 3D prints. Typically XCR is applied at around 300-500 grams per square meter; for smaller patterns you will need to factor a larger proportions to wastage (in the mixing cups brushes etc).

After coating the pattern will normally require flatting with 800 and 1200 grit [wet-and-dry abrasive paper](https://www.easycomposites.us/abrasive-paper) and then polishing with [NW1 cutting compound](https://www.easycomposites.us/nw1-white-super-cutting-compound).

- Mold Making Materials

In this project we are using a [high temperature gelcoat](https://www.easycomposites.us/eg160-high-temperature-epoxy-tooling-gel-coat) and [laminating paste](https://www.easycomposites.us/eg160-high-temperature-epoxy-tooling-gel-coat) which cures initially at room temperature but can be used in elevated temperature processes after releasing from the pattern. The mold making methods would be exactly the same if the molds were to be made from the standard temperature epoxy system [EG60](https://www.easycomposites.us/epoxy-tooling-gel-coat) and [EMP60](https://www.easycomposites.us/emp60-epoxy-molding-paste).

- Component Materials

The component is made using 1 ply of [XPREG XC110 210g 2/2 prepreg carbon fiber](https://www.easycomposites.us/xc110-210g-22-twill-3k-prepreg-carbon-fiber) with some localised areas with a second ply to increase strength where required. The vacuum bagging materials are [R210 unperforated release film](https://www.easycomposites.us/r210-unperforated-release-film), [BR180 breather cloth](https://www.easycomposites.us/br180-breather-fabric), [VB160 vacuum bagging film](https://www.easycomposites.us/vb160-vacuum-bagging-film), and [ST150 sealant tape](https://www.easycomposites.us/vacuum-bagging-sealant-tape).

- Prepreg Processing Equipment

For the vacuum bagging and vacuum hold during the cure we used the EC4 composites vacuum pump and the temperature controlled cure was done in our benchtop OV301 composites curing oven.
